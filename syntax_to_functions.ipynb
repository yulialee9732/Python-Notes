```python
# Set - unordered, changeable, no duplicates
thisSet = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}

thisSet.add("grape")            #adding an element
thisSet.update(tropical)        #merging tropical to thisSet
thisSet.remove("apple")         #removing apple - causes an error if not exist
thisSet.discard("tomato")       #removing apple - does NOT causes error if not exist
thisSet.clear()                 #clears the set
del thisSet                     #deletes the entire set

thisSet = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}
for x in thisSet:               #looping through each element in thisSet
    print(x)

newSet = thisSet.union(tropical) #returns new set with all items from both set & allows to combine a set with different data types, like list and tuple
newSet = thisSet | tropical     #same results as union

print(newSet)




```

```python
# Dictionaries - key:value pairs - ordered, changable, no duplicates for keys
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}

print(thisdict["brand"])            #prints value of brand
print(thisdict["model"])            #prints key of brand
print(len(thisdict))                #prints length of the dictionary
print(thisdict.items())             #prints each item in the dictionary



thisdict["color"] = "white"         #adding new item
x = thisdict.keys()
y = thisdict.values()

if "model" in thisdict:             #check if key exists
    print("True")

#Change and Update
thisdict["year"] = 2020             #making changes to original dictionary
thisdict.update({"owner": "me", "type": "fourDoor"}) #update allows multiple changes

#Remove
thisdict.pop("brand")               #removes an specified item
thisdict.popitem()                  #removes the last inserted item
del thisdict["model"]               #removes the specified key name
print(thisdict.keys())

#Loop
for x in thisdict:                  #prints all key names
    print(x)
for x in thisdict:                  #prints all values
    print(thisdict[x])
for x in thisdict.keys():           #prints all keys using keys() method
    print(x)
for x in thisdict.values():         #prints all values using values() method
    print(x)
for x,y in thisdict.items():
    print(x, y)

#copy
copy = thisdict.copy()              #copies a dictionary
cocopy = dict(thisdict)    #same thing
print(copy)
print(cocopy)

```

```python
# @decorator - lets you add additional behavior without changing the function's code

def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

@changecase
def otherfunction():
    return "hello my friends"

print(myfunction())
print(otherfunction())
```

```python
# decorator with argument
def myfunc(func):
    def myinner(x):
        return func(x).upper()
    return myinner

@myfunc
def lowerit(nam):
    return "Hello" + nam

print(lowerit("Yulia"))
```

```python
def multiply(a):
    return lambda b: a*b # CLOSURE — lambda captures 'a' from outer function
                         # and remembers it even after multiply() is done

double = multiply(2)  # double is now lambda b: 2*b — 'a' is locked in as 2
triple = multiply(3)  # triple is now lambda b: 3*b — 'a' is locked in as 3
quad   = multiply(4)  # quad   is now lambda b: 4*b — 'a' is locked in as 4

print(double(3))  # 2*3 = 6
print(triple(5))  # 3*5 = 15
print(quad(9))    # 4*9 = 36
```

```python
def fibonacci(n):
  if n <= 1:
    return n
  else:
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(7))
```

```python
#Recursion in function
def sum_function(numbers):
    if(len(numbers) == 0):
        return 0
    else:
        return numbers[0] + sum_function(numbers[1:])

mylist = [1,2,3,4,5]
print(sum_function(mylist))
```

```python
# Generator - functions that can be paused and resumed

def count_up_to(n):
  count = 1
  while count <= n:
    yield count
    count += 1

for num in count_up_to(5):
  print(num)
```
