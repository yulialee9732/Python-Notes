```python
# Dates

import datetime

print(datetime.datetime.now())
print(datetime.datetime.now().strftime("%B"))

# %a	Weekday, short version	Wed	
# %A	Weekday, full version	Wednesday	
# %w	Weekday as a number 0-6, 0 is Sunday	3	
# %d	Day of month 01-31	31	
# %b	Month name, short version	Dec	
# %B	Month name, full version	December	
# %m	Month as a number 01-12	12	
# %y	Year, short version, without century	18	
# %Y	Year, full version	2018	
# %H	Hour 00-23	17	
# %I	Hour 00-12	05	
# %p	AM/PM	PM	
# %M	Minute 00-59	41	
# %S	Second 00-59	08	
# %f	Microsecond 000000-999999	548513	
# %z	UTC offset	+0100	
# %Z	Timezone	CST	
# %j	Day number of year 001-366	365	
# %U	Week number of year, Sunday as the first day of week, 00-53	52	
# %W	Week number of year, Monday as the first day of week, 00-53	52	
# %c	Local version of date and time	Mon Dec 31 17:41:00 2018	
# %C	Century	20	
# %x	Local version of date	12/31/18	
# %X	Local version of time	17:41:00	
# %%	A % character	%	
# %G	ISO 8601 year	2018	
# %u	ISO 8601 weekday (1-7)	1	
# %V	ISO 8601 weeknumber (01-53)	01
```

```python
# Math (Built-in)
x=min(2,3,4,5)
print(x)

y = max(4,56,2,4)
print(y)

z=abs(-4)
print(z)

a=pow(2,4)
print(a)
```

```python
# Math Module
import math

x=math.sqrt(16)
print(x)

print(math.ceil(4.5))
print(math.floor(5.5))
print(math.pi)
```

```python
# JSON - exchanging and storing data
import json
x = {
  "name": "John",
  "age": 30,
  "city": "New York"
}

y=json.dumps(x)     #python -> json
print(y)
z=json.loads(y)     #json  -> python
print(z["name"])
```

```python
# TRY, EXCEPT, ELSE, FINALLY
# TRY: lets you test for errors
# EXCEPT: lets you handle the error
# ELSE: execute code when there is no error
# FINALLY: lets you execute code regardless of the results of try and except blocks

try:
  print(x)
except:
  print("Something went wrong")
finally:
  print("The 'try except' is finished")
```

```python
#String formatting
quantity = 3
itemno = 567
price = 49
myorder = "I want {} pieces of item number {} for {:.2f} dollars."
print(myorder.format(quantity, itemno, price))
```
