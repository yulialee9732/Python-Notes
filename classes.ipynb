```python
# Classes
class myClass:
    x=5

p1=myClass() #creating multiple objects from a class
p2=myClass()
p3=myClass()

print(p1.x)
print(p2.x)
print(p3.x)
```

```python
class Person:
    def __init__(self,name,age=25):
        self.name = name
        self.age = age

p1 = Person("Yulia")

print(p1.name, "is", p1.age)
```
