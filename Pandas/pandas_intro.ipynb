```python
# Series - in pandas, it's like a column of a table. One-dimensional array holding data
import pandas as pd

a=[1,4,6]
myvar = pd.Series(a)
newvar = pd.Series(a,index=["x","y","z"]) #creating labels

print(myvar)
print(myvar[0]) #indexing

print(newvar)
print(newvar["y"]) #indexing with label
```

```python
# Key/Value as Series
import pandas as pd

calories={"day1":180, "day2":300, "day3":400}
var=pd.Series(calories)
newvar=pd.Series(calories,index=["day1","day2"])

print(var)
print(newvar)
```

```python
# DataFrames - multi-dimensional tables
import pandas as pd

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
myvar = pd.DataFrame(data)

print(myvar)
print(myvar.loc[0]) #locating the first row
print(myvar.loc[[0,1]]) #locating the first and second row

newvar = pd.DataFrame(data,index=["day1","day2","day3"])
print(newvar)
print(newvar.loc["day2"])
```
