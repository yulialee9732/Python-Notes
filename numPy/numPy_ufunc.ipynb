```python
# Simple Arithmatic
import numpy as np

a=np.array([1,2,3,4,5])
b=np.array([6,7,8,9,10])

print("add:",np.add(a,b))
print("sum:",np.sum([a,b]))
print("sum with axis:",np.sum([a,b],axis=1)) #sum each array
print("cumsum a:",np.cumsum(a),"\n")

print("subtract:",np.subtract(a,b))
print("difference:",np.diff(a))
print("difference twice:",np.diff(a,n=2),"\n")

print("multiply:",np.multiply(a,b))
print("prod:",np.prod(a))
print("prod with axis:",np.prod([a,b],axis=1))
print("cumprod",np.cumprod(a))
print("divide:",np.divide(a,b))
print("power:",np.power(a,b),"\n")
print("mod:",np.mod(a,b)) #more commonly used
print("remainder:",np.remainder(a,b))
print("divmod:",np.divmod(b,a)) #b/a --> returns two arrays, quotients, and mod

c=np.array([-1,-5,3])

print("absolute:",np.absolute(c))
```

```python
# Rounding decimals - Truncation, rounding, floor, ceiling

import numpy as np

a=np.array([2.4364,3.1534,2.5,3.678,-2.26,-2.86,-2.5])

print(np.trunc(a))
print(np.around(a))
print(np.floor(a))
print(np.ceil(a))
```

```python
# Logs

import numpy as np

a=np.arange(1,10)
print(a)
print(np.log2(a))
print(np.log10(a))
print(np.log(a)) #natural log, e

# log at any base
from math import log
import numpy as np

nplog = np.frompyfunc(log, 2, 1)

print(nplog(100, 15))
```

```python
# LCM - Lowest Common Multiple
import numpy as np

a=4
b=3

print(np.lcm(a,b))
```

```python
# LCM on a array
import numpy as np

a=[1,2,3]

print(np.lcm.reduce(a))
```
