```python
import numpy as np

arr=np.array([1,2,3,4,5])
print(arr)
print(type(arr))
```

```python
# Creating arrays
'''
np.array([1, 2, 3])          # create array from list
np.zeros((3, 4))             # array filled with 0s
np.ones((3, 4))              # array filled with 1s
np.arange(0, 10, 2)         # [0, 2, 4, 6, 8] like range()
np.linspace(0, 1, 5)        # 5 evenly spaced numbers between 0 and 1
'''
```

```python
# Checking array info
'''
arr.dtype      # data type (int64, float64...)
arr.shape      # dimensions (rows, columns)
arr.ndim       # number of dimensions
arr.size       # total number of elements
'''
```

```python
# Reshaping
'''
arr.reshape(2, 3)    # change shape without changing data
arr.flatten()        # convert any dimension → 1D
arr.transpose()      # flip rows and columns
'''
```

```python
# math operations
'''
np.sum(arr)          # total sum
np.mean(arr)         # average
np.median(arr)       # median
np.min(arr)          # minimum value
np.max(arr)          # maximum value
np.std(arr)          # standard deviation
'''
```

```python
# Joining & Splitting
'''
np.concatenate((arr1, arr2))    # join arrays
np.vstack((arr1, arr2))         # stack vertically (row-wise)
np.hstack((arr1, arr2))         # stack horizontally (column-wise)
np.array_split(arr, 3)          # split into 3 parts
'''
```

```python
# Searching & Sorting
'''
np.sort(arr)                    # sort array
np.where(arr > 3)               # find index where condition is true
np.searchsorted(arr, 5)         # find index where 5 should be inserted
'''
```

```python
# Type Conversion
'''
arr.astype(int)      # convert to int
arr.astype(float)    # convert to float
arr.astype(str)      # convert to string
'''
```
