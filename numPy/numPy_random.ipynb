```python
# loc - (Mean) where the peak of the bell exists.
# scale - (Standard Deviation) how flat the graph distribution should be.
# size - The shape of the returned array.

from numpy import random
x=random.normal(size=(2,3))
print(x)
```

```python
from numpy import random
x=random.normal(loc=1, scale=2, size=(2,3))
print(x)
```

```python
# visualization of normal distribution
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.normal(size=1000), kind="kde")
plt.show()
```

```python
# Binomial Distribution
# It is a discrete distribution, describing the outcome being yes/no question.
from numpy import random
import seaborn as sns
import matplotlib.pyplot as plt

sns.displot(random.binomial(n=10,p=0.5,size=10)) #n:number of trials, p: possibility, size:number of experiment of that trials

plt.show()
```

```python
# Poisson Distribution
# It's a discrete distribution. It estimates how many times it will happen in specified time.

from numpy import random
import seaborn as sns
import matplotlib.pyplot as plt

sns.displot(random.poisson(lam=2, size=1000))

plt.show()
```

```python
# Uniform Distribution
# Probability where every event has equal chances of happening
# three parameters: low, high, size
#       low - lower bound, defualt to 0.0
#       high - higher bound, default to 1.0
#       size - shape of the returned array

from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.uniform(size=1000), kind="kde")

plt.show()
```

```python
# Logistic Distribution
# predicting whether something will happen or not (yes/no)
# extensively used in Machine Learning - Logistic Regression
# three parameters: loc, scale, size
#       loc - mean, where the peak is, default 0
#       scale - standard deviation, the flatness of distribution, default 1
#       size - shape of the returned array


from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
sns.displot(random.logistic(size=1000),kind="kde")
plt.show()
```

```python
# Multinomial Distribution
# It's a generalization of binomial distribution - more than two choice
# three parameters: n, pvals, size
#       n - number of experiment ran
#       pvals - list of probabilities of outcomes (e.g. [1/6, 1/6, 1/6, 1/6, 1/6, 1/6] for dice roll).
#       size - shape of the returned array

from numpy import random

x=random.multinomial(n=6, pvals=[1/6,1/6,1/6,1/6,1/6,1/6])
print(x)
```

```python
# Exponential Distribution
# It's used for describing time until next event e.g. succes/failure etc.
# two parameters: scale, size
#       scale - inverse of rate. default to 0
#       size - shape of returned array

from numpy import random

x=random.exponential(scale=2, size=(2,3))
print(x)
```

```python
from numpy import random
import seaborn as sns
import matplotlib.pyplot as plt

sns.displot(random.exponential(size=1000), kind="kde")
plt.show()
```

```python
#  Chi Square Distribution
# It's used for comparsion between the prediction and actual result
# two parameters: df, size
#       df - degree of freedom
#       size - shape of returned array

from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.chisquare(df=1, size=1000), kind="kde")

plt.show()
```

```python
# Rayleigh Distribution
# Speed or distance when two independent forces combine
# two parameters: scale, size
#       scale - (standard deviation) decides how flat the distribution will be. default 1.0
#       size - shape of returned array

from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.rayleigh(size=1000), kind="kde")

plt.show()
```

```python
# Pareto Distribution
# A distribution following pareto's law. 80-20 distribution (20% factors cuases 80%)
# Pareto → WHAT PROPORTION do the top few control?
# two parameters: a, size
#       a - shape parameter
#       size - shape of returned array

from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.pareto(a=2, size=1000))

plt.show()
```

```python
# Zipf Distribution
# Zipf  → HOW MUCH does each rank get?
# two parameters: a, size
#       a - distribution parameter
#       size - shape of returned array

from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

x = random.zipf(a=2, size=1000)
sns.displot(x[x<10])

plt.show()
```
