```python
# open file using open()
# r for Read - error if the file does not exist
# a for append - opens a file for appending, creates one if not exist
# w for write - opens a file for writing, creates one if not exist
# x for create - creates the specified file, returns an error if the file exist

# t for text file
# b for binary file

# f = open("demofile.txt") same as below
# f = open("demofile.txt", "rt") rt is the default
```

```python
# Manual open and close. Not commonly used. error arises if the file is not closed
f = open("file_handling/demofile.txt")
print(f.read())
f.close()
```

```python
# w/ WITH statement
with open("file_handling/demofile.txt") as f:
    print(f.read())
```

```python
# w/ WITH statement first five characters
with open("file_handling/demofile.txt") as f:
    print(f.read(5))
```

```python
# w/ WITH statement one line
with open("file_handling/demofile.txt") as f:
    print(f.readline())
```

```python
# Read each line
with open("file_handling/demofile.txt") as f:
    for x in f:
        print(x)
```

```python
# "a" - Append - will append to the end of the file
# "w" - Write - will overwrite any existing content
```

```python
with open("file_handling/demofile.txt", "a") as f:
    f.write("Now it has more content!") 
```

```python
with open("file_handling/demofile.txt") as f:
    print(f.read())
```

```python
with open("demofile.txt", "w") as f:
  f.write("Woops! I have deleted the content!")

#open and read the file after the overwriting:
with open("demofile.txt") as f:
  print(f.read())
```

```python
# "x" - Create - will create a file, returns an error if the file exists
# "a" - Append - will create a file if the specified file does not exists
# "w" - Write - will create a file if the specified file does not exists
```

```python
# Deleting a file
import os
os.remove("file_handling/demofile.txt")
```
